```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT_DIR = Path().cwd().parent
DATASET = ROOT_DIR / "data" / "diamonds.csv"
ARTIFACTS_PATH = Path().cwd().parent / "artifacts"
df = pd.read_csv(DATASET,encoding="utf-8")

df_clean = df.copy()
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
positive_cols = [
    "carat", "price", "x", "y", "z",
    "depth", "table"
]

invalid_mask = df_clean[positive_cols].le(0).any(axis=1)
df_clean = df_clean.loc[~invalid_mask]
```

```python
valid_categories = {
    "cut": [
        "Fair", "Good", "Very Good",
        "Premium", "Ideal"
    ],
    "color": list("DEFGHIJ"),
    "clarity": [
        "I1", "SI2", "SI1", "VS2",
        "VS1", "VVS2", "VVS1", "IF"
    ]
}
```

```python
for col in valid_categories:
    df_clean[col] = df_clean[col].astype("string").str.strip()
```

```python
from pathlib import Path

output_path = Path(ROOT_DIR / "data/processed/diamonds_clean.csv")

output_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

df_clean.to_csv(output_path, index=False)
```
